```python
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['order_approved_at'] = pd.to_datetime(df['order_approved_at'])
df['order_delivered_carrier_date'] = pd.to_datetime(df['order_delivered_carrier_date'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])

df.info()
```

```python
query = "SELECT * FROM orders_master LIMIT 5"
pd.read_sql(query, conn)
```

```python
query = """
SELECT strftime('%Y-%m', order_purchase_timestamp) as month, COUNT(order_id) as total_orders
FROM orders_master
GROUP BY month
ORDER BY month
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT customer_id, SUM(price) as total_spent, COUNT(DISTINCT order_id) as total_orders
FROM orders_master
GROUP BY customer_id
ORDER BY total_spent DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT total_orders, COUNT(customer_id) as num_customers
FROM (
    SELECT customer_id, COUNT(DISTINCT order_id) as total_orders
    FROM orders_master
    GROUP BY customer_id
)
GROUP BY total_orders
ORDER BY total_orders
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT payment_type, COUNT(DISTINCT order_id) as total_orders, AVG(payment_value) as avg_payment_value
FROM orders_master
GROUP BY payment_type
ORDER BY total_orders DESC
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
    strftime('%Y-%m', order_purchase_timestamp) as month,
    SUM(price) as monthly_revenue
FROM orders_master
GROUP BY month
ORDER BY month
"""
revenue_trend = pd.read_sql(query, conn)
revenue_trend['growth_rate_%'] = revenue_trend['monthly_revenue'].pct_change() * 100
revenue_trend
```

```python
query2 = """
SELECT
    strftime('%Y-%m', order_purchase_timestamp) as month,
    SUM(price) as monthly_revenue,
    SUM(freight_value) as monthly_freight_cost,
    (SUM(freight_value) * 100.0 / SUM(price)) as freight_pct_of_revenue
FROM orders_master
GROUP BY month
ORDER BY month
"""
pd.read_sql(query2, conn)
```
